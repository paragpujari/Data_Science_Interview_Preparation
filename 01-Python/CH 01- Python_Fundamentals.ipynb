##### Important concepts to understand

- Variable          : A name referring to a value.
- Data type         : Identifies the kind of value, such as an integer or string.
- Literal           : A value written directly in the code.
- Operator          : A symbol or keyword used to perform an operation.
- Expression        : A combination of values, variables and operators that produces a result.
- Type conversion   : Converting a value from one data type to another.
- Input             : Receiving information from the user.
- Output            : Displaying information using print().
- Indentation       : Leading whitespace that Python uses to structure code blocks.

15. Write a Python program to calculate the total bill for three products using their prices and quantities, then calculate the average cost per product unit. Identify the data types used

```python
print('-----------------------------First Product---------------------------------------------------------------------')
product1  = 'Laptop'
price1    =  87567.56
quantity1 =  2
bill1 = price1 * quantity1
print(f'First bill is {bill1}')

print('-----------------------------Second Product---------------------------------------------------------------------')
product2  = 'Smartphones'
price2    =  77567.34
quantity2 =  3
bill2 = price2 * quantity2
print(f'Second bill is {bill2}')

print('-----------------------------Third Product---------------------------------------------------------------------')
product3  = 'Earphones'
price3    =  76567.66
quantity3 =  5
bill3 = price3 * quantity3
print(f'Third bill is {bill3}')

##### Calculate the total bill
total_bill = bill1 + bill2 + bill3
print(f'Total bill of all the items is {total_bill}')

##### Total product units
total_quantity = quantity1 + quantity2 + quantity3
print(f'Total quantity is {total_quantity}')

##### Compute the average cost per product unit
average_cost = total_bill/total_quantity
print(f'Average cost per product unit is {average_cost:.2f}')
```

16. Write a Python program to accept two numbers and display their sum, difference, product, quotient and remainder. Handle the case where the second number is zero before division

```python
a = int(input("Enter the first number"))
b = int(input("Enter the second number"))

sum=a+b
diff=a-b
pro=a*b

print(f'Sum value is {sum}')
print(f'Diff value is {diff}')
print(f'Pro value is {pro}')

if(b==0):
    print(f'Quotient cannot be calculated')
    print(f'Remainder cannot be calculated')
else:
    quo = a%b
    rem = a/b
    print(f'Quotient value is {quo}')
    print(f'Remainder value is {rem}')
```

18. Write a Python program to convert a total number of seconds into hours, minutes and remaining seconds using arithmetic operators

```python
total_seconds = 3675

###### Convert the total seconds into hours
hours = total_seconds//3600
print(f'Hours is {hours}')
##### Convert the total seconds into minutes
minutes = (total_seconds%3600)//60
print(f'Minutes is {minutes}')
##### Convert the total seconds into remaining seconds
rem_sec = total_seconds%60
print(f'Remaining seconds is {rem_sec}')
```

19. Write a Python program to accept the principal amount, annual interest rate and time in years, then calculate simple interest and the total amount payable

```python
principal = int(input("Enter the principal"))
rate      = int(input("Enter the rate"))
time      = int(input("Enter the time"))

##### Calculate the simple interest
si = (principal * rate * time)/100.0
print(f'Simple Interest is {si}')

##### Calculate the amount
total_amount = principal + si
print(F'Total amount is {total_amount}')
```

20. Write a Python program to accept a three-digit integer and calculate the sum of its digits using arithmetic operators, without converting the integer to a string

```python
num = int(input("Enter the three digit integer"))
s=0
x=num
while(x!=0):
    d=x%10
    s=s+d
    x=x//10
print(f'Sum of the digits is {s}')
```
